```python
import pandas as pd
import seaborn as sns
import statistics as sts
```

```python
dataset = pd.read_csv( r'C:\Users\guilherme_doge1\Desktop\ciência de dados\Aula 05 - 14-09-2026\Atividade 1\tempo.csv',
                      sep=";")
```

```python
agrupado_aparencia = dataset.groupby(["Aparencia"]).size()
agrupado_aparencia
```

```python
agrupado_aparencia.plot.bar()
```

```python
dataset.loc[dataset["Aparencia"] == "menos", "Aparencia"] = "sol"
```

```python
agrupado_temperatura = dataset.groupby(["Temperatura"]).size()
agrupado_temperatura
```

```python
agrupado_temperatura.plot.bar()
```

```python
sns.boxplot(dataset["Temperatura"]).set_title("Temperatura")
```

```python
dataset["Temperatura"] = dataset["Temperatura"].astype(float)
mediana_temperatura = sts.median(dataset["Temperatura"])
dataset.loc[dataset["Temperatura"] > 130, "Temperatura"] = mediana_temperatura
```

```python
dataset.loc[dataset["Temperatura"] > 130, "Temperatura"]
```

```python
dataset.isnull().sum()
```

```python
dataset["Umidade"] = dataset["Umidade"].fillna(sts.median(dataset["Umidade"]))
```

```python
dataset.isnull().sum()
```

```python
agrupado_vento = dataset.groupby(["Vento"]).size()
agrupado_vento
```

```python
dataset.loc[~dataset["Vento"].isin(["VERDADEIRO", "FALSO"]), "Vento"] = "FALSO"
```
