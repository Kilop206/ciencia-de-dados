```python
# Mantém apenas o primeiro registro de cada motor repetido
df.drop_duplicates(subset="Id", keep="first", inplace=True)
```

```python
# Corrige erros de digitação nas unidades
df.loc[df["Unidade"].isin(["JSS"]), "Unidade"] = "JS"
df.loc[df["Unidade"].isin(["SCH"]), "Unidade"] = "SC"
df.loc[df["Unidade"].isin(["GUARAMIRIM"]), "Unidade"] = "GM"
```

```python
# Valores sem correspondência e nulos recebem a categoria mais frequente
df.loc[df["Unidade"].isin(["SP", "XX"]), "Unidade"] = "JS"
df.loc[df["Unidade"].isnull(), "Unidade"] = "JS"
```

```python
# Padroniza os nomes dos tipos de motor
df.loc[df["Tipo_Motor"].isin(["M", "Mono"]), "Tipo_Motor"] = "Monofásico"
df.loc[df["Tipo_Motor"].isin(["3-Fase", "Trif", "TRIF"]), "Tipo_Motor"] = "Trifásico"
df.loc[df["Tipo_Motor"].isnull(), "Tipo_Motor"] = "Trifásico"
```

```python
# Padroniza o estado de conexão
df.loc[df["Status_Conexao"].isin(["OFF", "0"]), "Status_Conexao"] = "Desconectado"
df.loc[df["Status_Conexao"].isin(["OK", "1"]), "Status_Conexao"] = "Conectado"
df.loc[df["Status_Conexao"].isnull(), "Status_Conexao"] = "Conectado"
```

```python
# Para dados numéricos ausentes, usa a mediana porque os valores extremos afetam menos o resultado
df.loc[df["Temperatura"].isnull(), "Temperatura"] = 75.11
df.loc[df["Horas_Operacao"].isnull(), "Horas_Operacao"] = 47.3
df.loc[df["Pressao_Oleo"].isnull(), "Pressao_Oleo"] = 3.76
df.loc[df["Vibracao"].isnull(), "Vibracao"] = 1.78
df.loc[df["Custo_Manutencao"].isnull(), "Custo_Manutencao"] = 10723.055
```

```python
# Valores fisicamente impossíveis também são substituídos pela mediana
df.loc[(df["Temperatura"] > 120) | (df["Temperatura"] < 0), "Temperatura"] = 75.11
df.loc[(df["Horas_Operacao"] > 100) | (df["Horas_Operacao"] < 0), "Horas_Operacao"] = 47.3
df.loc[df["Pressao_Oleo"] <= 0, "Pressao_Oleo"] = 3.76
df.loc[df["Vibracao"] < 0, "Vibracao"] = 1.78
```
